```python
# -------------------------------------------------------------------------------------------------------------- #
#                                                                                                                #
#   Import necessary packages                                                                                    #
#                                                                                                                #
# -------------------------------------------------------------------------------------------------------------- #

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
from pathlib import Path


# -------------------------------------------------------------------------------------------------------------- #
#                                                                                                                #
#   Define the EXPERIMENTS for every experiment                                                                  #
#                                                                                                                #
# -------------------------------------------------------------------------------------------------------------- #

# EXPERIMENTS = [
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_003',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_004',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_005',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_006',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_007',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_008',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_009',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_010',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_011',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_012',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\test_anormal_experiment_013',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_001\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_002\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_002\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_002\\test_anormal_experiment_003',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_002\\test_anormal_experiment_004',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_002\\test_anormal_experiment_005',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_002\\test_anormal_experiment_006',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_003',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_004',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_005',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_006',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_007',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_008',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\test_anormal_experiment_009',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_003\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_004\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_004\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_004\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_005\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_006\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_007\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_007\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_008\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_008\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_009\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_009\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_010\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_010\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_011\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_011\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_012\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_012\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_013\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_013\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_014\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_014\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_014\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_014\\train_normal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_015\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_015\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_016\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_016\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_017\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_017\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_018\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_018\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_019\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_020\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_021\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_021\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_022\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_022\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_023\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_023\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_024\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_025\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_026\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_026\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_026\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_027\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_027\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_028\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_028\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_028\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_029\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_029\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_030\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_030\\test_anormal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_030\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_031\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_031\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_031\\train_normal_experiment_002',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_032\\test_anormal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_033\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_034\\train_normal_experiment_001',
#     'batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_035\\train_normal_experiment_001',]

# EXPERIMENTS = [
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_001\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_001\\test_anormal_experiment_002',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_002\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_002\\train_normal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_003\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_003\\train_normal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_004\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_004\\train_normal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_005\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_005\\test_anormal_experiment_002',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_006\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_006\\train_normal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_007\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_007\\train_normal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_008\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_009\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_009\\train_normal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_010\\train_normal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_011\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_011\\test_anormal_experiment_002',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_012\\test_anormal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_012\\test_anormal_experiment_002',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_012\\train_normal_experiment_001',
#     'batch_dist_ternary_acetone+butan-1-ol+methanol\\operating_point_013\\test_anormal_experiment_001',
#     'batch_dist_binary_ethanol+propan-2-ol\\operating_point_001\\test_anormal_experiment_001',
#     'batch_dist_binary_ethanol+propan-2-ol\\operating_point_001\\test_anormal_experiment_002',
#     'batch_dist_binary_ethanol+propan-2-ol\\operating_point_002\\train_normal_experiment_001',
#     'batch_dist_binary_ethanol+propan-2-ol\\operating_point_002\\train_normal_experiment_002']

# EXPERIMENTS = ['batch_dist_binary_ethanol+propan-2-ol\\operating_point_002\\train_normal_experiment_002']                      # single experiment for testing purposes
EXPERIMENTS = ['batch_dist_ternary_butan-1-ol+propan-2-ol+water\\operating_point_013\\test_anormal_experiment_001']                      # single experiment for testing purposes

# -------------------------------------------------------------------------------------------------------------- #
#                                                                                                                #
#   For-Loop to iterate through all experiments                                                                  #
#                                                                                                                #
# -------------------------------------------------------------------------------------------------------------- #

for PATH in EXPERIMENTS:


    # -------------------------------------------------------------------------------------------------------------- #
    #                                                                                                                #
    #   Define the File to extract                                                                                   #
    #                                                                                                                #
    # -------------------------------------------------------------------------------------------------------------- #

    ANOMALY_META_YAML = f"../data/raw//00_Timeseries_Label_Anomaly_Metadata/Operation/{PATH}.yaml"
    OPERATION_ANOMALY = f"../data/raw//00_Timeseries_Label_Anomaly_Metadata/Operation/{PATH}.csv"
    OPERATION_SENSORS = f"../data/raw/01_Timeseries_Sensors/Operation/{PATH}.csv"
    STARTUP_SENSORS = f"../data/raw/01_Timeseries_Sensors/Startup/{PATH}.csv"
    SHUTDOWN_SENSOR = f"../data/raw/01_Timeseries_Sensors/Shutdown/{PATH}.csv"
    OPERATION_ACTUATORS = f"../data/raw/02_Timeseries_Actuators/Operation/{PATH}.csv"
    STARTUP_ACTUATORS = f"../data/raw/02_Timeseries_Actuators/Startup/{PATH}.csv"
    SHUTDOWN_ACTUATORS = f"../data/raw/02_Timeseries_Actuators/Shutdown/{PATH}.csv"

    # Single time use
    # PLANT_DESCRIPTION_YAML = f"../data/raw/06_Tabular_Plant_Information/{PATH}/{MAPPING}.yaml"
    PLANT_SETTINGS = f"../data/raw/06_Tabular_Plant_Information/{PATH}/settings.csv"


    # -------------------------------------------------------------------------------------------------------------- #
    #                                                                                                                #
    #   Path for processed plots                                                                                     #
    #                                                                                                                #
    # -------------------------------------------------------------------------------------------------------------- #

    output_dir = Path(r"../data/processed/plots")
    output_dir.mkdir(parents=True, exist_ok=True)


    # -------------------------------------------------------------------------------------------------------------- #
    #                                                                                                                #
    #   Read the Files                                                                                               #
    #                                                                                                                #
    # -------------------------------------------------------------------------------------------------------------- #

    # with open(ANOMALY_META_YAML, "r", encoding="utf-8") as file:
    #     anomaly_meta = yaml.safe_load(file)

    operation_anomaly = pd.read_csv(OPERATION_ANOMALY)
    operation_sensors = pd.read_csv(OPERATION_SENSORS)
    startup_sensors = pd.read_csv(STARTUP_SENSORS)
    shutdown_sensors = pd.read_csv(SHUTDOWN_SENSOR)
    operation_actuators = pd.read_csv(OPERATION_ACTUATORS)
    startup_actuators = pd.read_csv(STARTUP_ACTUATORS)
    shutdown_actuators = pd.read_csv(SHUTDOWN_ACTUATORS)

    # with open(PLANT_DESCRIPTION_YAML, "r", encoding="utf-8") as file:
    #     plant_description = yaml.safe_load(file)

    plant_settings = pd.read_csv(PLANT_SETTINGS)


    # -------------------------------------------------------------------------------------------------------------- #
    #                                                                                                                #
    #   Merge sensors and actuators time series (startup - operation - shutdown) included as 'phase'                 #
    #                                                                                                                #
    # -------------------------------------------------------------------------------------------------------------- #

    exp_startup = pd.merge(
        startup_sensors,
        startup_actuators, 
        on='Time', 
        how='inner'
    )
    exp_startup['phase']="startup"


    exp_operation = pd.merge(
        operation_sensors,
        operation_actuators, 
        on='Time', 
        how='inner'
    )

    exp_operation['phase']="operation"


    exp_shutdown = pd.merge(
        shutdown_sensors,
        shutdown_actuators, 
        on='Time', 
        how='inner'
    )
    exp_shutdown['phase']="shutdown"


    # -------------------------------------------------------------------------------------------------------------- #
    #                                                                                                                #
    #   Concatenate startup - operation - shutdown                                                                   #
    #                                                                                                                #
    # -------------------------------------------------------------------------------------------------------------- #

    experiment = pd.concat(
        [
            exp_startup, 
            exp_operation, 
            exp_shutdown
        ], 
        ignore_index=True
    )


    # -------------------------------------------------------------------------------------------------------------- #
    #                                                                                                                #
    #   Insert Meta Data - Batch, Operating Point, Test, elapsed Time, and Anomaly Label                             #
    #                                                                                                                #
    # -------------------------------------------------------------------------------------------------------------- #

    BATCH = PATH.split('\\')[0]
    OPERATING_POINT = PATH.split('\\')[1]
    TEST = PATH.split('\\')[2]

    experiment['batch'] = BATCH
    experiment['operating_point'] = OPERATING_POINT
    experiment['experiment'] = TEST
    experiment.head()

    time = pd.to_timedelta(experiment["Time"])

    experiment.insert(
        experiment.columns.get_loc("Time") + 1,
        "elapsed_time",
        (time - time.iloc[0]).dt.total_seconds()
    )

    experiment = pd.merge(
        experiment,
        operation_anomaly,
        on='Time',
        how='left'
    )

    experiment = experiment.rename(columns={'Label (anomaly)': 'anomaly_label'})
    experiment.loc[experiment['anomaly_label'].isna(), 'anomaly_label'] = 0



    # -------------------------------------------------------------------------------------------------------------- #
    #                                                                                                                #
    #   Handling corrupted data                                                                                      #
    #                                                                                                                #
    # -------------------------------------------------------------------------------------------------------------- #

    # ============================================================
    # CORRECT CORRUPTED VALUES
    # ============================================================

    # Sensors where values > 500 are considered corrupted
    columns_over_500 = [
        "T701",
        "T702",
        "T703",
        "T704",
        "T706",
        "T708",
        "T709",
        "T711",
        "T712",
        "T705",
        "PY23",
        "TV1",
        "H702",
        "H706",
        "H708",
        "H704",
        "P701",
        "P702"
    ]

    # Sensors where values > 1 are considered corrupted
    columns_over_1 = [
        "FT703",
        "FT704",
        "PDI701",
        "PDI702"
    ]


    for column in columns_over_500:
        mask = experiment[column] > 500
        experiment.loc[mask, column] /= 1000

    for column in columns_over_1:
        mask = abs(experiment[column]) > 1
        experiment.loc[mask, column] /= 1000


    # -------------------------------------------------------------------------------------------------------------- #
    #                                                                                                                #
    #   Plotting the anomaly and temperatures                                                                        #
    #                                                                                                                #
    # -------------------------------------------------------------------------------------------------------------- #

    # ============================================================
    # SETTINGS
    # ============================================================

    temperatures = [
        'T701',
        'T702',
        'T703',
        'T704',
        'T705',
        'T706',
        'T708',
        'T709',
        'T711',
        'T712'
    ]

    pressures = [
        'PDI701',
        'PDI702',
        'PY23'
    ]

    level_sensors = [
        'LS701',
        'LS702'
    ]

    distillate = [
        'P701',
        'P702',
        'FT703',
        'FT704'
    ]

    heating_sections = [
        'H701',
        'H702',
        'H706',
        'H704',
        'H708'
    ]

    vacuum_system = [
        'P301',
        'TV1'
    ]

    cooling_system = [
        'FYI702',
        'AV716'
    ]

    nitrogen = [
        'AV709'
    ]

    other_features = [
        'AV8'
    ]

    # ============================================================
    # FIND PHASE BOUNDARIES
    # ============================================================

    phase_change = experiment["phase"].ne(
        experiment["phase"].shift()
    )

    phase_indices = experiment.index[phase_change]

    # ============================================================
    # CREATE FIGURE
    # ============================================================

    fig, axes = plt.subplots(
        nrows=9,
        ncols=1,
        figsize=(14, 30),
        sharex=True,
        gridspec_kw={
            "height_ratios": [
                1,      # anomaly
                3,      # temperatures
                2,      # pressures
                2,      # level sensors
                2,      # distillate
                2,      # heating
                2,      # vacuum
                2,      # cooling
                2       # nitrogen + other
            ]
        }
    )

    ax_label = axes[0]
    ax_temp = axes[1]
    ax_pressure = axes[2]
    ax_level = axes[3]
    ax_distillate = axes[4]
    ax_heating = axes[5]
    ax_vacuum = axes[6]
    ax_cooling = axes[7]
    ax_other = axes[8]

    # ============================================================
    # 1. ANOMALY LABEL
    # ============================================================

    sns.lineplot(
        data=experiment,
        x="elapsed_time",
        y="anomaly_label",
        ax=ax_label
    )

    ax_label.set_ylabel("Anomaly")
    ax_label.set_xlabel("")

    # ============================================================
    # 2. TEMPERATURES
    # ============================================================

    # T702-T712 on primary axis
    for feature in temperatures:

        if feature != "T701":

            sns.lineplot(
                data=experiment,
                x="elapsed_time",
                y=feature,
                ax=ax_temp,
                label=feature
            )

    ax_temp.set_ylabel("Temperature")
    ax_temp.set_xlabel("")
    ax_temp.set_title("Temperature Sensors")

    # T701 on secondary axis
    ax_temp_secondary = ax_temp.twinx()

    sns.lineplot(
        data=experiment,
        x="elapsed_time",
        y="T701",
        ax=ax_temp_secondary,
        label="T701"
    )

    ax_temp_secondary.set_ylabel("T701")

    # Combine legends from both temperature axes
    lines_1, labels_1 = ax_temp.get_legend_handles_labels()
    lines_2, labels_2 = ax_temp_secondary.get_legend_handles_labels()

    ax_temp.legend(
        lines_1 + lines_2,
        labels_1 + labels_2,
        ncol=2,
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    # Remove automatic secondary legend
    ax_temp_secondary.get_legend().remove()

    # ============================================================
    # 3. PRESSURES
    # ============================================================

    for feature in pressures:

        sns.lineplot(
            data=experiment,
            x="elapsed_time",
            y=feature,
            ax=ax_pressure,
            label=feature
        )

    ax_pressure.set_ylabel("Pressure")
    ax_pressure.set_xlabel("")
    ax_pressure.set_title("Pressure Sensors")

    ax_pressure.legend(
        ncol=2,
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    # ============================================================
    # 4. LEVEL SENSORS
    # ============================================================

    for feature in level_sensors:

        sns.lineplot(
            data=experiment,
            x="elapsed_time",
            y=feature,
            ax=ax_level,
            label=feature
        )

    ax_level.set_ylabel("Level")
    ax_level.set_xlabel("")
    ax_level.set_title("Level Sensors")

    ax_level.legend(
        ncol=2,
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    # ============================================================
    # 5. DISTILLATE
    # ============================================================

    for feature in distillate:

        sns.lineplot(
            data=experiment,
            x="elapsed_time",
            y=feature,
            ax=ax_distillate,
            label=feature
        )

    ax_distillate.set_ylabel("Value")
    ax_distillate.set_xlabel("")
    ax_distillate.set_title("Distillate")

    ax_distillate.legend(
        ncol=2,
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    # ============================================================
    # 6. HEATING SECTIONS
    # ============================================================

    for feature in heating_sections:

        sns.lineplot(
            data=experiment,
            x="elapsed_time",
            y=feature,
            ax=ax_heating,
            label=feature
        )

    ax_heating.set_ylabel("Value")
    ax_heating.set_xlabel("")
    ax_heating.set_title("Heating Sections")

    ax_heating.legend(
        ncol=2,
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    # ============================================================
    # 7. VACUUM SYSTEM
    # ============================================================

    for feature in vacuum_system:

        sns.lineplot(
            data=experiment,
            x="elapsed_time",
            y=feature,
            ax=ax_vacuum,
            label=feature
        )

    ax_vacuum.set_ylabel("Value")
    ax_vacuum.set_xlabel("")
    ax_vacuum.set_title("Vacuum System")

    ax_vacuum.legend(
        ncol=2,
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    # ============================================================
    # 8. COOLING SYSTEM
    # ============================================================

    for feature in cooling_system:

        sns.lineplot(
            data=experiment,
            x="elapsed_time",
            y=feature,
            ax=ax_cooling,
            label=feature
        )

    ax_cooling.set_ylabel("Value")
    ax_cooling.set_xlabel("")
    ax_cooling.set_title("Cooling System")

    ax_cooling.legend(
        ncol=2,
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    # ============================================================
    # 9. NITROGEN + OTHER
    # ============================================================

    for feature in nitrogen + other_features:

        sns.lineplot(
            data=experiment,
            x="elapsed_time",
            y=feature,
            ax=ax_other,
            label=feature
        )

    ax_other.set_ylabel("Value")
    ax_other.set_xlabel("Time")
    ax_other.set_title("Nitrogen / Other")

    ax_other.legend(
        ncol=2,
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

    # ============================================================
    # PHASE BOUNDARIES
    # ============================================================

    for start_idx in phase_indices:

        phase = experiment.loc[start_idx, "phase"]
        phase_time = experiment.loc[start_idx, "elapsed_time"]

        if start_idx != experiment.index[0]:

            for ax in axes:

                ax.axvline(
                    phase_time,
                    linestyle="--",
                    alpha=0.5
                )

            # Secondary temperature axis
            ax_temp_secondary.axvline(
                phase_time,
                linestyle="--",
                alpha=0.5
            )

        # Phase label
        for ax in axes:

            ax.text(
                phase_time,
                1.02,
                phase,
                transform=ax.get_xaxis_transform(),
                ha="left"
            )

    # ============================================================
    # TITLE
    # ============================================================

    fig.suptitle(
        f"{BATCH} | {OPERATING_POINT} | {TEST}",
        fontsize=14
    )

    # ============================================================
    # LAYOUT
    # ============================================================

    plt.tight_layout()

    # ============================================================
    # SAVE
    # ============================================================

    plt.savefig(
        rf"{output_dir}/{BATCH}/{OPERATING_POINT}_{TEST}_anomaly_temp_ts.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)
```
